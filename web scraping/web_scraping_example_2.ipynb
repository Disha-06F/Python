##### <span style="color:orange"> requests:</span> For sending web requests.
#####  <span style="color:orange">BeautifulSoup:</span> For parsing and extracting HTML data.
#####  <span style="color:orange">pandas:</span> For saving the data into a structured format (CSV file).

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

##### <span style="color:orange">requests.get(url)</span> sends a GET request to the website.
##### <span style="color:orange">response.raise_for_status()</span> checks for errors (like 404 or 500).
##### If the request fails, it raises an exception instead of silently failing

```python
url="https://books.toscrape.com/"
response=requests.get(url)
response
```

```python
response.raise_for_status()
```

##### Converts the raw HTML code <span style="color:orange">(response.text)</span>into a BeautifulSoup object.
#####  <span style="color:orange"> html.parser </span>lets BeautifulSoup understand the HTML structure.
##### Now soup acts like a navigable tree of the page’s HTML.

```python
soup=BeautifulSoup(response.text,"html.parser")
```

```python
soup
```

##### Looks for all<span style="color:orange"> <article> </span>tags with the class "product_pod".
##### Each <span style="color:orange">"product_pod"</span> represents one book on the page.
##### books is now a list of BeautifulSoup objects — one per book.

```python
books = soup.find_all("article",class_ ="product_pod")
```

```python
books
```

#####  <span style="color:orange">book.h3.a["title"]:</span> Gets the title attribute from the <a> tag inside <h3> — that’s the book title.
##### <span style="color:orange">book.find("p", class_="price_color").text:</span> Finds the price paragraph (<p class="price_color">) and extracts its text — book price.
##### <span style="color:orange">book.find("p", class_="instock availability").text.strip(): </span>Finds the availability info (like “In stock”) and removes extra spaces.

```python
data=[]
for book in books:
    title=book.h3.a["title"]
    price=book.find("p",class_="price_color").text
    stock=book.find("p",class_="instock availability").text.strip()

    data.append({
       "Title":title,
       "Price":price,
       "Availability":stock
    })
```

```python
data
```

##### Converts the list of dictionaries into a pandas DataFrame — a table with columns Title, Price, and Availability.

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
df.to_csv("books.csv",index=False)
```
