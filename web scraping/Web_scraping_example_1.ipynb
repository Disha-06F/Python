#### <span style="color:Red">requests →</span> Used to send HTTP requests to websites and get their HTML content.
#### <span style="color:Red">BeautifulSoup (from bs4) →</span> Used to parse and extract data from HTML.
#### <span style="color:Red">pandas →</span> Used for creating and saving structured data (like CSV files)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

##### Defines the URL to scrape.
##### <span style="color:Red">requests.get(url) </span>sends a GET request to fetch the HTML code of the page.
##### response contains the full server response, including the page’s HTML content and status code.
##### If response.status_code == <span style="color:Red">200</span>, the request was successful.

```python
url = "https://quotes.toscrape.com/"
response=requests.get(url)
response
```

##### Takes the HTML text from the response <span style="color:Red">(response.text)</span>
##### Parses it using Python’s built-in <span style="color:Red">"html.parser"</span>, turning it into a structured tree that’s easy to navigate.

```python
soup=BeautifulSoup(response.text,"html.parser")
```

```python
soup
```

##### <span style="color:Red"> soup.find_all() </span>searches the parsed HTML for specific<span style="color:Red"> tags and classes.</span>
##### quotes → a list of all <span class="text"> elements (each contains a quote).
##### authors → a list of all <small class="author"> elements (each contains the author’s name).

```python
quotes =soup.find_all("span",class_="text")
authors=soup.find_all("small",class_="author")
```

```python
quotes
```

```python
authors
```

##### <span style="color:Red"> data = [] </span>Creates an empty list to store all the extracted quote–author pairs as dictionaries.
##### <span style="color:Red"> zip(quotes, authors) →</span> pairs each quote with its author (so they can be processed together).
##### For every pair:
##### quote.text extracts just the text inside the <span> tag.
##### author.text extracts just the text inside the <small> tag
##### These are stored in a <span style="color:Red"> dictionary</span> like:
#### <span style="color:Blue"> {"Quote": "The world as we have created it...", "Author": "Albert Einstein"}</span>
##### <span style="color:Red"> .append() </span>adds this dictionary to the data list.

```python
data=[]
for quote,author in zip(quotes,authors):
    data.append({
        "Quote":quote.text,
        "Author":author.text
    })
```

```python
data
```

#### Converts the list of dictionaries (data) into a pandas DataFrame — basically a<span style="color:Red">  table</span> with two columns: Quote and Author.

```python
df=pd.DataFrame(data)
```

```python
df
```

#### Exports the DataFrame to a file named quotes.csv.
#### <span style="color:Red"> index=False</span> means the row numbers (0, 1, 2, …) won’t be saved as a separate column.

```python
df.to_csv("quotes.csv",index=False)
```
